## Big picture
The bigram model sees one previous character. Here we see $C$ previous characters, each mapped to a learned vector
(an embedding), concatenated and passed through one tanh layer. Embeddings are the first thing the Transformer also
has (§3.4); everything after them will change.

| step | shape |
|---|---|
| X (token ids) | (N, C) |
| E[X] | (N, C, d) |
| flatten | (N, C·d) |
| a = flat @ W1 + b1, h = tanh(a) | (N, H) |
| logits = h @ W2 + b2 | (N, V) |
