## Big picture
The bigram table stored probabilities directly. Neural models output unnormalized scores (logits) $z \in \mathbb{R}^V$;
softmax turns them into a distribution, and cross-entropy scores that distribution against the true next token.
This is P1's sigmoid and binary cross-entropy, generalized from 2 classes to V.

Shapes: logits $(N, V)$, targets $(N,)$. For a sequence model, logits are $(B, T, V)$: reshape to $(B \cdot T, V)$ first.
