## Big picture
A language model assigns a probability to the next token. The simplest one: count how often each character follows
each other character, and normalize. Its loss, the mean negative log-likelihood (NLL, natural log, in nats per
character), is the number every later model in this repo must beat. Reference point: a uniform guess over $V$
characters scores $\ln V$.
