## Big picture
Characters make sequences long; whole words make vocabularies explode and leave unknown words. BPE starts from small
units and repeatedly merges the most frequent adjacent pair. Starting from UTF-8 bytes means any string in any script
can be encoded (outside knowledge: GPT-2's byte-level BPE).

**Why this matters for Khmer (interpretation).** Khmer characters (U+1780 to U+17FF) take 3 bytes each in UTF-8, and
Khmer writing does not put spaces between words. A tokenizer trained mostly on English spends many more tokens per
Khmer character, which makes Khmer sequences longer and more expensive. Measure it instead of assuming it.
